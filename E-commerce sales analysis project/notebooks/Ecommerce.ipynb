```python
# Remove duplicate rows from the DataFrame in-place
df.drop_duplicates(inplace=True)
```

```python
# Replace specific string values ('N/A', 'NULL', empty string) with NaN across the DataFrame
df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
# Display the entire DataFrame after cleaning
df
```

```python
# Strip leading/trailing whitespace from string (object) columns
df = df.apply(lambda x: x.str.strip() if x.dtype =='object' else x)
```

```python
# Convert 'Customer_Name', 'City', and 'State' columns to title case
df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
# Remove rows where the 'Email' column does not contain '@'
df = df[df['Email'].str.contains('@',na=False)]
```

```python
# Convert 'Order_Date' and 'Delivery_Date' columns to datetime objects, coercing errors to NaT
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
# Convert 'Qty', 'Unit_Price', and 'Discount' columns to numeric, coercing errors to NaN
cols = ['Qty', 'Unit_Price', 'Discount']

for c in cols:
  df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
# Filter out rows where 'Qty' is not greater than 0
df = df[df['Qty']>0]
```

```python
# Fill missing 'Discount' values with 0
df['Discount'] = df['Discount'].fillna(0)

# Fill missing 'Phone' values with 'Unknown'
df['Phone'] = df['Phone'].fillna('Unknown')

# Fill missing 'Delivery_Date' values with the corresponding 'Order_Date'
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

**Feature Engineering**

```python
#Total sales
df['sales']=df['Qty']*df['Unit_Price']
```

```python
#net amount

df['Net_Amount'] = df['sales'] - (df['sales'] * df['Discount']/100)
```

```python
#Profit (example: 20%)
df['Profit'] = df['Net_Amount'] * 0.20
```

```python
#order Month

df['Month']=df['Order_Date'].dt.month_name()
```

```python
#order year

df["Year"] = df["Order_Date"].dt.year
```

```python
#weekday

df['Weekday']=df['Order_Date'].dt.day_name()
```
